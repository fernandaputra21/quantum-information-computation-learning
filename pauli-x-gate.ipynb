# Pauli X-Gate

```python
from qiskit import *
from qiskit.visualization import plot_bloch_multivector, visualize_transition, plot_histogram
from qiskit_aer import Aer
from qiskit.quantum_info import Statevector
```

```python
# Create single qubit quantum circuit
qc = QuantumCircuit(1) # 1 number of qubit, by default |0> or [1,0]
```

```python
initial_state = [0,1]

qc.initialize(initial_state)
```

```python
initial_state = Statevector.from_instruction(qc)

print("Initial State:")
plot_bloch_multivector(initial_state)
```

```python
# Apply pauli x-gate

qc.x(0)

qc.draw('mpl') # Draw circuit
```

```python
after_state = Statevector.from_instruction(qc)

print("after State:")
plot_bloch_multivector(after_state)
```

```python
# Animated Transition
visualize_transition(qc)
```

```python
qc.measure_all()
# Simulate a real quantum measurement
# Open Quantum Assembly Language - QASM

backend = Aer.get_backend('qasm_simulator')

job = backend.run(qc, shots = 1024)
counts = job.result().get_counts()
plot_histogram(counts)
```
